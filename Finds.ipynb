```python
import csv
n = 6
a=[]

print('The given training ')
with open('Find_S_Dataset.csv', 'r') as file:
    csvreader = csv.reader(file)
    for row in csvreader:
        a.append(row)
        print(row)
```

```python
print('')
hypothesis = ['0']*n
print('Hypothesis: ', hypothesis)

for j in range(0,n):
  hypothesis[j] = a[0][j]

for i in range(0,len(a)):
  if a[i][n] == 'yes':
    for j in range(0,n):
      if a[i][j] != hypothesis[j]:
        hypothesis[j] = '?'
      else:
        hypothesis[j] = a[i][j]

  print(f'For training instance no:{i} the hypothesis is', hypothesis)

print('\n The maximally specific hypothesis for given traning examples: ')
print(hypothesis)
```
